## 2. Target or proxy

## 3. Success metric

*One metric you can defend. What number means 'good'?*

**Answer:**

    The main success metric for this task is **NDCG (Normalized Discounted Cumulative Gain)**
    because it evaluates how well the model ranks the most important content pages near the top of the list.
    A good ranking model should prioritize pages with the highest optimization potential,
    allowing the content team to focus their efforts where they will have the greatest impact.

```python
success_metric = "NDCG (Normalized Discounted Cumulative Gain)"
print("Success Metric:", success_metric)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

**Answer:**

    A fixed rule cannot accurately determine which content pages should be prioritized because search performance depends on multiple factors that interact with each other.
    For example, a page with a low CTR but very high search volume may have a greater optimization opportunity than a page with a slightly lower ranking but very little traffic. Similarly, average position, competition,
    impressions, and content freshness all influence the priority of a page.
    A machine learning model can learn these complex relationships from historical data and produce a better ranking of pages than a single manually defined rule. This allows the content team to focus on pages with the greatest potential impact.
